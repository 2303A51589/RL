```python
# Policy Iteration Algorithm in a Deterministic 2x2 Grid World

# 1. Define MDP
states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.0001

normal_reward = -1
wall_reward = -1
goal_reward = 100

# Grid Layout:
# S0 | Wall
# S1 | G

transitions = {
    'S0': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward)
    },

    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward)
    }
}

# 2. Initialize Policy and State Values
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Display MDP Information
print("POLICY ITERATION ALGORITHM")
print("-" * 50)

print("\nStates:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

print("\nTransition Information:")
for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

print("\nInitial Policy:", policy)

# 3. Policy Iteration
iteration = 0

while True:

    iteration += 1

    print("\nIteration:", iteration)

    # 4. Policy Evaluation
    while True:

        delta = 0

        for state in ['S0', 'S1']:

            old_value = V[state]

            action = policy[state]

            next_state, reward = transitions[state][action]

            V[state] = reward + gamma * V[next_state]

            delta = max(delta, abs(old_value - V[state]))

        if delta < theta:
            break

    print("\nPolicy Evaluation:")
    print("S0 =", round(V['S0'], 4))
    print("S1 =", round(V['S1'], 4))
    print("G  =", V['G'])

    # 5. Policy Improvement
    policy_stable = True

    old_policy = policy.copy()

    for state in ['S0', 'S1']:

        action_values = {}

        for action in actions:

            next_state, reward = transitions[state][action]

            q_value = reward + gamma * V[next_state]

            action_values[action] = q_value

        best_action = max(action_values, key=action_values.get)

        policy[state] = best_action

    print("\nImproved Policy:", policy)

    # 6. Check Policy Stability
    if old_policy == policy:
        policy_stable = True
    else:
        policy_stable = False

    if policy_stable:
        print("\nPolicy is Stable!")
        break

# 7. Calculate Final Action Values
print("\n" + "=" * 50)
print("FINAL OPTIMAL RESULTS")
print("=" * 50)

print("\nOptimal State Values:")

for state in states:
    print(state, "=", round(V[state], 4))

print("\nFinal Action-Value Table:")

for state in ['S0', 'S1']:

    print("\nState:", state)

    q_values = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        q_value = reward + gamma * V[next_state]

        q_values[action] = q_value

        print(action, "=", round(q_value, 4))

# 8. Display Optimal Policy
print("\nOptimal Policy:")

for state in ['S0', 'S1']:
    print(state, "->", policy[state])

print("G -> Terminal")

# 9. Bellman Optimality Verification
print("\nBellman Optimality Verification:")

for state in ['S0', 'S1']:

    q_values = []

    for action in actions:

        next_state, reward = transitions[state][action]

        q_value = reward + gamma * V[next_state]

        q_values.append(q_value)

    max_q = max(q_values)

    print("\nState:", state)
    print("V*(s) =", round(V[state], 4))
    print("Max Q*(s,a) =", round(max_q, 4))

    if abs(V[state] - max_q) < theta:
        print("Bellman Equation Verified!")

# 10. Display Optimal Path
print("\nOptimal Path:")

current_state = 'S0'

while current_state != 'G':

    print(current_state, end=" -> ")

    action = policy[current_state]

    next_state, reward = transitions[current_state][action]

    current_state = next_state

print("G")
print("\nPolicy Iteration Completed Successfully!")
```
