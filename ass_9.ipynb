```python
# Assignment 1: Algebraic Solution of Bellman Expectation Equations

gamma = 0.9

# Solve the simultaneous equations:
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

# Calculate action values
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

# Verify state values using the policy
V_S0_check = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V_S1_check = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

# Display results
print("ASSIGNMENT 1: ALGEBRAIC SOLUTION")
print("--------------------------------")
print(f"V(S0) = {V_S0:.2f}")
print(f"V(S1) = {V_S1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up) = {Q_S1_Up:.2f}")

print("\nVerification:")
print(f"V(S0) from action values = {V_S0_check:.2f}")
print(f"V(S1) from action values = {V_S1_check:.2f}")
```
